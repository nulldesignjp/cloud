````python
import os
import sys

def get_runtime_env():
    # 1. Google Colabの判定
    if 'google.colab' in sys.modules:
        return 'colab'

    # 2. Vertex AI Workbench (Managed/User-managed) の判定
    # Vertex AIのインスタンスには特有の環境変数が設定されています
    if os.getenv('DL_ANACONDA_HOME') or os.getenv('XDG_CONFIG_HOME') == '/home/jupyter/.config':
        return 'vertex_ai'

    # 3. Anaconda / ローカル環境の判定
    if 'conda' in sys.version or os.path.exists(os.path.join(sys.prefix, 'conda-meta')):
        return 'anaconda'

    return 'local_python'

# 使用例
env = get_runtime_env()
print(f"現在の実行環境: {env}")

if env == 'colab':
    from google.colab import auth
    auth.authenticate_user()
elif env == 'vertex_ai':
    print("Vertex AI環境のため、追加認証なしでCloudリソースにアクセスを試みます")
````

````python
from google.colab import drive
import os

# Googleドライブのマウント
try:
    drive.mount('/content/drive')
    print('Googleドライブが正常にマウントされました。')
except Exception as e:
    print(f'Googleドライブのマウント中にエラーが発生しました: {e}')

# Google Driveのルートディレクトリ
drive_root = '/content/drive/My Drive'

# ドライブがマウントされているか再確認し、ファイル一覧を表示
if os.path.exists(drive_root):
    print(f"\n{drive_root} 内のファイルとフォルダ一覧:")
    for item in os.listdir(drive_root):
        print(item)
else:
    print(f"Google ドライブが {drive_root} にマウントされていません。ファイル一覧を表示できません。")
````

````python
import os

drive_root = '/content/drive/My Drive'
print(f"\n{drive_root} の内容 (ツリー表示):\n")

if os.path.exists(drive_root):
    # drive_rootのパスセグメントの数を取得し、これを基準の深さとする
    drive_root_segments = drive_root.split(os.sep)
    base_depth = len(drive_root_segments) - (1 if drive_root.startswith(os.sep) else 0)

    for dirpath, dirnames, filenames in os.walk(drive_root):
        # 現在のディレクトリパスの深さを計算
        current_segments = dirpath.split(os.sep)
        current_depth = len(current_segments) - (1 if dirpath.startswith(os.sep) else 0)

        # drive_rootからの相対的な深さを計算
        relative_depth = current_depth - base_depth

        # インデントとプレフィックスを設定
        indent = '    ' * relative_depth

        # ディレクトリ名を表示
        if dirpath == drive_root:
            print(f"{drive_root}")
        else:
            # 最後のディレクトリの場合、'└── 'を使い、それ以外は'├── 'を使う
            # ここではシンプルに'├── 'を使用
            print(f"{indent}├── {os.path.basename(dirpath)}{os.sep}")

        # サブディレクトリを表示
        for dirname in dirnames:
            print(f"{indent}│   ├── {dirname}{os.sep}")

        # ファイルを表示
        for filename in filenames:
            print(f"{indent}│   └── {filename}")

else:
    print(f"Google ドライブが {drive_root} にマウントされていません。ファイル一覧を表示できません。")
````
